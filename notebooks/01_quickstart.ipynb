# DriftScope — Quickstart

End-to-end walkthrough: fetch CMEMS currents → run particles → visualize.

Make sure `.env` has your Copernicus Marine credentials before starting.

````python
%load_ext autoreload
%autoreload 2

from datetime import date, timedelta
from driftscope.config import PRESETS, SimConfig
from driftscope.fetch import fetch_currents
from driftscope.simulate import run_simulation
from driftscope.viz import plot_static, load_trajectories, traj_to_dataframe
````

## 1. Pick a region

````python
region = PRESETS['sundarbans']
print(region)
print('bbox:', region.bbox)
````

## 2. Fetch CMEMS currents (2 weeks)

````python
end = date.today() - timedelta(days=2)
start = end - timedelta(days=14)

nc_path = fetch_currents(
    bbox=region.bbox,
    start_date=start.isoformat(),
    end_date=end.isoformat(),
    region_name=region.name,
)
````

## 3. Seed & advect 500 particles

````python
cfg = SimConfig(
    n_particles=500,
    runtime_days=10,
    seed_lon=89.2,   # offshore Sundarbans
    seed_lat=21.4,
    seed_radius_deg=0.15,
)
traj_path = run_simulation(nc_path, cfg)
````

## 4. Plot

````python
plot_static(traj_path)
````
